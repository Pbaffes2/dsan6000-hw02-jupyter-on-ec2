```python
import boto3
import os
```

```python
s3 = boto3.client("s3")

s3.create_bucket(Bucket="dsan6000-pb1100")
```

```python
# Checking if created

s3.list_buckets()
```

```python
for file in os.listdir("data/"):
    if file.endswith(".parquet"):
        local_path = os.path.join("data", file)
        key = f"wikipedia-hourly/{file}"
        s3.upload_file(Filename = local_path,Bucket = "dsan6000-pb1100", Key = key)
```

```python
# Check if it added it to s3 bucket

bucket = "dsan6000-pb1100"
prefix = "wikipedia-hourly"

s3 = boto3.client("s3")

if prefix and not prefix.endswith("/"):
    prefix += "/"

paginator = s3.get_paginator("list_objects_v2")
counter = 0
for page in paginator.paginate(Bucket= bucket, Prefix = prefix):
    for item in page["Contents"]:
        if item["Key"].endswith(".parquet"):
            print(item["Key"])
            counter += 1
print(f"Number of files is {counter}")
```
